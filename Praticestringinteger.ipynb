```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b) #union
```

```python
print(a & b) #Intersection
```

```python
print(a - b) #Difference
```

```python
print(b.issubset(a))  # True
print(a.issuperset(b))  # True
```

```python
a | b    # union
a & b    # intersection
a - b    # difference
a ^ b    # symmetric difference
```

```python
a ^ b 
```
